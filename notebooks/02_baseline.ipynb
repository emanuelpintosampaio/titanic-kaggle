# Titanic - Baseline e primeira submissão

Antes de usar machine learning, uma previsão bem simples serve de referência: **toda mulher sobrevive, todo homem morre**. Qualquer modelo depois precisa superar essa nota.

## 1. Carregando os dados de teste

```python
import pandas as pd

test = pd.read_csv("../data/test.csv")
test.head()
```

```python
# (linhas, colunas)
test.shape
```

São **418 passageiros** e **11 colunas**: as mesmas do treino, menos `Survived`, que é o que precisamos prever.

## 2. Previsão: mulher sobrevive

```python
# True/False para "é mulher?", convertido em 1/0
test['Survived'] = (test['Sex'] == 'female').astype(int)

# Conferindo: todo female deve ter 1 e todo male, 0
test[['Sex', 'Survived']]
```

### Quanto essa regra acertaria no treino?

No `train.csv` sabemos a resposta certa, então dá para medir a **acurácia** da regra: a proporção de passageiros em que a previsão bate com o que aconteceu.

```python
train = pd.read_csv("../data/train.csv")

previsao_treino = (train['Sex'] == 'female').astype(int)
(previsao_treino == train['Survived']).mean()
```

A regra acerta **78,7%** no treino. É o número a bater.

## 3. Montando a submissão

```python
# O Kaggle espera exatamente duas colunas: PassengerId e Survived
submissao = test[['PassengerId', 'Survived']]
submissao
```

### Conferências antes de enviar

```python
# 418 linhas e 2 colunas
submissao.shape
```

```python
# Nenhum valor faltando
submissao.isnull().sum()
```

```python
# Quantos sobreviventes previstos (1) e quantos não (0)
submissao['Survived'].value_counts()
```

## 4. Salvando o arquivo

```python
# index=False evita uma coluna extra com o número da linha
submissao.to_csv('../submissions/baseline_sexo.csv', index=False)
```
